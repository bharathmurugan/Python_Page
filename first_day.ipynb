Cloud SLA Uptime Refund Calculator
Company: Google Cloud / Walmart Cloud Infrastructure
Topic: Multi-Branch if-elif-else & Funnel Architecture

Scenario:
Enterprise cloud providers guarantee Service Level Agreements (SLAs) for server uptime. If server downtime breaches SLA thresholds, enterprise clients are awarded SLA penalty refunds calculated against their monthly bill.

Refund Policy Thresholds:

Uptime $\ge$ 99.9%: No refund (0%)
Uptime $\ge$ 98.0% and < 99.9%: 10% refund on monthly bill
Uptime < 98.0%: 25% refund on monthly bill

Code Optimization Rule:
Store the calculated value in a single variable (refund) inside the conditional branches, and write a single output print(refund) statement at the bottom of your script (Funnel Pattern).


```python
uptime=97.5
bill=50000
if uptime >=99.9:
  refund=0
elif uptime>=98:
  refund=bill*0.10
else:
  refund=bill*0.25
print("Refund amount:",refund)        #this is bottle neck problem -after complete each code this will move to the last print statement.
```

```python

python = {"rama", "ravi", "praduman", "london", "patlu"}
sql = {"rama", "krishna", "sita", "radha"}
java = {"rama", "kiwi", "king authors", "coffee bean", "krishna"}

print("Subscribe to Python channel:", python - sql - java)

print("People subscribed to more than 1 channel:",
      (python & sql) | (sql & java) | (python & java))

print("People subscribed to all 3 channels:",
      python & sql & java)

print("People subscribed to exactly 1 channel:",
      (python - sql - java) |
      (sql - python - java) |
      (java - python - sql))

a = (python & sql) | (sql & java) | (python & java)

# Remove people who subscribe to all 3 channels
two_subscribe = a - (python & sql & java)

print("Exactly two subscriptions:", two_subscribe,
      "count:", len(two_subscribe))

all_members = python | sql | java

not_two = all_members - two_subscribe

print("Do not have exactly two subscriptions:", not_two,
      "count:", len(not_two))

print("Total members:", all_members)
print("Total member count:", len(all_members))
```

**List and Dict Comprehension**

**Split function:**

Split is a function used to convert a larger into smaller words by default it will take space and meta characters(special characters)


```python
print("Split function:")
for x in "rama,is,going" .split(","):
  print(len(x),x)

#list
print("Using List:")
print([len(x) for x in "rama,is,going" .split(",")])
#dict
print("Using Dict:")
print({x:len(x) for x in "rama,is,going" .split(",")})
```

**Last rule:**

 In the comprehension if we have written both if and else block then we need to write before for loop.

 If condition expression should be comes first.




```python
for x in [18,90,12,44,18]:
  if x>18:
    print("allowed")
  else:
    print("not allowed")

print("list comprehension:")

#using list comprehension
print(["allowed" if x>18 else "not allowed" for x in [18,90,12,44,18]])

print("Dictionary comprehension:")

#using dictionary
{x:"allowed" if x>18 else "not allowed" for x in [18,90,12,44,18]}
```

```python
#using dict comprehension- it will get int datatype
print({x:x**2 for x in range(5) if x%2==0})
```

# Object Specifications

## 1. Instagram

### Properties
* **username**: Unique string identifier for the account.
* **followersCount**: Integer tracking the number of followers.
* **posts**: List of media objects (photos, videos, reels).
* **feedAlgorithm**: System logic that dictates content ranking.
* **isPrivate**: Boolean indicating account privacy status.

### Behavior
* **uploadPost()**: Publishes a photo or video to the grid.
* **likeMedia()**: Registers an engagement on another user's post.
* **sendDirectMessage()**: Transmits text or media privately to a user.
* **refreshFeed()**: Fetches and displays new content based on the algorithm.

---

## 2. Blinkit

### Properties
* **deliveryAddress**: Physical location coordinates or string for the drop-off.
* **cartItems**: List of products selected by the user.
* **deliveryPartner**: Object containing details of the assigned rider.
* **estimatedDeliveryTime**: Integer representing minutes until arrival.
* **paymentStatus**: Enum representing the state of payment (Pending, Paid, Failed).

### Behavior
* **addItemToCart()**: Adds a specific inventory item to the active cart.
* **placeOrder()**: Finalizes the cart and initiates the transaction process.
* **trackLiveLocation()**: Streams real-time GPS coordinates of the delivery rider.
* **processRefund()**: Reverses a transaction for missing or damaged items.

---

## 3. College

### Properties
* **collegeName**: String representing the name of the institution.
* **departments**: List of academic branches (e.g., Computer Science, Mechanical).
* **studentBody**: Collection of all enrolled student objects.
* **facultyList**: Collection of all employed professor objects.
* **accreditationStatus**: Grade or ranking tier assigned by education boards.

### Behavior
* **enrollStudent()**: Admits a new student and generates a unique student ID.
* **conductExamination()**: Schedules and executes assessments across departments.
* **generateReportCard()**: Calculates CGPA/grades and issues official transcripts.
* **organizePlacementDrive()**: Hosts companies and coordinates recruitment events.


Class:

class is the blueprint of a object

Object:

Object is the real world entity.

Note:

Using one blueprint can create multiple of objects.







```python
class Car:
    def __init__(self):
        self.brand = 'Lamborghini Aventador'   #self is internal purpose
        self.color = 'yellow'
        self.cost = '8.70 crs'

c1 = Car() #outer communication

# Print its individual properties
print("Brand:", c1.brand)
print("Color:", c1.color)
print("Cost:", c1.cost)
```

```python
class Car:
    def __init__(self):
        self.brand = 'Lamborghini Aventador'   #self is internal purpose
        self.color = 'yellow'
        self.cost = '8.70 crs'
    def start_engine(self):
        print(self.brand,"is starting")
    def shift_gear(self):
        print(self.brand,"is shifting gear")
    def accelerate(self):
        print(self.brand,"is accelerating")

c1 = Car() #outer communication

# Print its individual properties
print("Brand:", c1.brand)
print("Color:", c1.color)
print("Cost:", c1.cost)

c1.start_engine()
c1.shift_gear()
c1.accelerate()
```
